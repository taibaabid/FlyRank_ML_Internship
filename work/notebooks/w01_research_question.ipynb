## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

###** Problem Framing  **

**The Core Question:**
Which high-exposure pages are showing the strongest signs of performance decay (decline in clicks, impressions, or engagement) and should be reviewed first for a content refresh?

**The Unit of Analysis:**
A single pseudonymized content item (URL/page level) per client.

**The Decision It Improves:**
It optimizes editorial resource allocation by shifting from random or strictly calendar-based content updates to an evidence-backed, high-priority review schedule.

**The Concrete Action:**
A content reviewer or editor opens the top-ranked pages in the queue, inspects the associated reason codes, and manually updates, expands, or optimizes the content.

**The Cost of a Wrong Call:**
A False Positive costs manual labor—an editor wastes valuable time reviewing a page that doesn't actually need adjustments. A False Negative costs lost revenue and traffic—a major traffic-driving page continues to silently decay without intervention.

**Why ML/Data is Needed:**
Simple heuristics (like checking age alone) ignore complex combinations of traffic momentum, search volume, and engagement signals across thousands of URLs.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Load the small anonymized starter dataset
df = pd.read_csv('content_refresh_anonymized.csv')
# Calculate 3 real numbers to justify the project
total_rows = len(df)
declining_count = len(df[df['trend_direction'] == 'down'])
pct_declining = (declining_count / total_rows) * 100
avg_impressions = df['impressions_90d'].mean()

print(f"--- Dataset Quick Metrics ---")
print(f"1. Total content items in starter dataset: {total_rows:,}")
print(f"2. Content items currently in active decline: {declining_count:,} ({pct_declining:.2f}%)")
print(f"3. Average impressions over prior 90 days: {avg_impressions:,.2f}")
```
